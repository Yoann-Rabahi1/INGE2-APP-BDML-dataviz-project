```python
def count_plot(df, col, fig_size=(8, 6)):

    plt.figure(figsize=fig_size)
    sns.countplot(
        data=df,
        x=f"{col}"
    )

    plt.title(f"Répartition des différents {col}")
    plt.xlabel(f"{col}")
    plt.show()    
```

```python
count_plot(df, "weekday")
```

```python
count_plot(df, "Base")
```

```python
count_plot(df, "hour")
```
